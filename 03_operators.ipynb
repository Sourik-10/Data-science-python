```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

```python
x = 5
x += 3
print("x += 3 →", x)

x -= 2
print("x -= 2 →", x)

x *= 4
print("x *= 4 →", x)

x /= 2
print("x /= 2 →", x)
```

```python
a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
x = True
y = False

print("AND:", x and y)
print("OR:", x or y)
print("NOT x:", not x)
```

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a is not c)
```

```python
nums = [10, 20, 30, 40]

print(20 in nums)
print(50 not in nums)
```

```python
a = 5   # 0101
b = 3   # 0011

print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```
