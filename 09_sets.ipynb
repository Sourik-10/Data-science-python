```python
nums = {10, 20, 30}

nums.remove(20)   # error if not present
nums.discard(50)  # no error
nums.pop()        # random element

nums
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference A-B:", A - B)
print("Difference B-A:", B - A)
print("Symmetric Difference:", A ^ B)
```

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

print(A.issubset(B))
print(B.issuperset(A))
print(A.isdisjoint({7, 8}))
```
