| Priority | Operators              |   |
| -------- | ---------------------- | - |
| 1️⃣      | `()`                   |   |
| 2️⃣      | `**`                   |   |
| 3️⃣      | `+  -  ~` (unary)      |   |
| 4️⃣      | `*  /  //  %`          |   |
| 5️⃣      | `+  -`                 |   |
| 6️⃣      | `<<  >>`               |   |
| 7️⃣      | `&`                    |   |
| 8️⃣      | `^`                    |   |
| 9️⃣      | `                      | ` |
| 🔟      | `==  !=  >  <  >=  <=` |   |
| 11️⃣     | `not`                  |   |
| 12️⃣     | `and`                  |   |
| 13️⃣     | `or`                   |   |
| 14️⃣     | `=  +=  -=  *=  /=`    |   |


```python
result = 10 + 3 * 2
print(result)
```

```python
result = 10 + 3 * 2 ** 2
print(result)
```

```python
result = (10 + 3) * 2
print(result)
```
