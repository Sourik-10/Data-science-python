```python
nums = [10, 20, 30, 40, 50]

nums.remove(30)   # value দিয়ে
nums.pop()        # last element
nums.pop(1)       # index দিয়ে
del nums[0]

print(nums)
```

```python
nums = [1, 2, 3, 4, 5]

squares = [x**2 for x in nums]
even = [x for x in nums if x % 2 == 0]

print(squares)
print(even)
```

```python
nums = [5, 1, 9, 3]

print(max(nums))
print(min(nums))
print(sum(nums))

nums.sort()
print(nums)

nums.reverse()
print(nums)
```
